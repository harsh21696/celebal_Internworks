```python
# importing pandas :
import pandas as pd
```

```python
# loading the dataset :
df = pd.read_csv("Combined_dataset.csv")
print("Dataset Loaded Successfully!")
```

```python
# displaying the first 5 rows of the dataset :
df.head()
```

```python
# displaying the last 5 rows of the dataset :
df.tail()
```

```python
# displaying the number of rows and columns in the dataset :
print("Rows and Columns:", df.shape)
```

```python
# displaying the column names in the dataset :
print(df.columns)
```

```python
# displaying the data types of each column in the dataset :
print(df.dtypes)
```

```python
# displaying the information of the dataset :
df.info()
```

```python
# checking for missing values in the dataset :
df.isnull().sum()
```

```python
# filling missing values in the dataset :
df["discount"] = df["discount"].fillna(0)
df["seller_name"] = df["seller_name"].fillna("Unknown")
df["seller_information"] = df["seller_information"].fillna("Not Available")
```

```python
# dropping the "videos" column from the dataset :
df.drop("videos", axis=1, inplace=True)
```

```python
# verifying missing values in the dataset :
df.isnull().sum()
```

```python
# selecting specific columns to display :
df[["title", "rating", "final_price"]].head()
```

```python
# filtering the dataset to include only products with a rating greater than 4 :
high_rating = df[df["rating"] > 4]
high_rating.head()
```

```python
# filtering the dataset to include only products with a final price greater than 500 :
expensive_products = df[pd.to_numeric(df["final_price"].str.replace('₹', '').str.replace(',', ''), errors='coerce') > 500]
expensive_products.head()
```

```python
# checking for duplicate rows in the dataset :
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# removing duplicate rows from the dataset :
df.drop_duplicates(inplace=True)
print("Duplicates Removed")
```

```python
# creating a new column "quantity" and filling it with the value 1 for all rows :
df["quantity"] = 1
```

```python
#creating a new column "total_amount" by multiplying "final_price" and "quantity" columns :
df["total_amount"] = df["final_price"] * df["quantity"]
df[["final_price", "quantity", "total_amount"]].head()
```

```python
# saving the cleaned dataset to a new CSV file :
df.to_csv("cleaned_dataset.csv", index=False)
print("Cleaned Dataset Saved Successfully!")
```
